# The Analysis of Shopping Sites
## Prediction of User Intent on Online Shopping Sites

### Problem Description

Online shopping has become the epitome of convenience in the 21th century, with millions of users inter-
acting with diverse online shopping sites every day. In this project, we aim to predict a user’s purchase
intent by modeling the sequence of behaviors of them when browsing an online shopping sites. Since the
market of online shopping is extremely competitive due to the abundance of online shopping platforms,
accurately inferring purchase intent from noisy, sequential browsing behaviors can significantly improve
conversion rate, recommendation quality, and overall user experience on shopping platforms.

### Dataset

- Source: Roman Zykov, Noskov Artem, and Anokhin Alexander. (2022). Retailrocket recommender
system dataset [Data set]. Kaggle. https://doi.org/10.34740/KAGGLE/DSV/4471234
- Contents: The dataset contains behavioral data (`clicks`, `add-to-cart`, and `purchases`) from a real-
world e-commerce website. The events.csv contains over 2.7 million events from 1.4 million unique
visitors over 4.5 months.
- Key Features: {Category Id, Parent Id, Visitor Id, Item Id, Property, Event}
